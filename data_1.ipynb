```python
import pandas as pd
from pathlib import Path
```

```python
base_dir = Path(r"C:\Users\тема\Desktop")
train_src = base_dir / "combined_dataset_emotions.csv" 
train_dst = base_dir / "train_prepared.csv"
```

```python
# Колонки эмоций
emotion_cols = ["Joy", "Sadness", "Surprise", "Fear", "Anger"]

def prepare_dataset(path_in, path_out):
    df = pd.read_csv(path_in)

    if "dataset" in df.columns:
        df = df.drop(columns=["dataset"])

    row_sums = df[emotion_cols].sum(axis=1)

    bad_mask = row_sums != 1
    problem_count = bad_mask.sum()

    if problem_count > 0:
        df_clean = df[~bad_mask].copy()
    else:
        df_clean = df.copy()

    def get_label(row):
        for i, col in enumerate(emotion_cols):
            if row[col] == 1:
                return i
        return -1

    df_clean["label"] = df_clean.apply(get_label, axis=1)

    df_out = df_clean[["text", "label"]].copy()

    df_out.to_csv(path_out, index=False)
    print(f" Файл сохранён: {path_out}")
    
    print("\nПервые 5 строк результата:")
    print(df_out.head())

prepare_dataset(train_src, train_dst)
```
